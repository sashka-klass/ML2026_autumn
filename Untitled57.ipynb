#Глубина вывода
1)При каком ранге визуальное качество перестает отличаться от оригинала?


Ранг должен быть очень высоким, близким к рангу матрицы

2)Для какого ранга достигается сжатие в 10 раз?

при ранге от 10 до 30, примерно 17-18

3)Почему для темных/светлых участков ошибка разная?

в темных участках нужно больше сингулярных чисел, принизком ранге они страдают больше, тк берется меньше сингулярных чисел, на светлых это не так заметно
